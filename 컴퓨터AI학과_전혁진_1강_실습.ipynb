```python
#1.리스트(List)
fruits = ["apple", "banana","banana" ,"strawberry","apple",]

print(fruits[2])
fruits[2]="peach"
print(fruits[2])
print(fruits)
```

```python

#2.튜플(Tuple)
fruits = ("apple", "banana","banana" ,"strawberry","apple",)
print(fruits[2])
print(fruits)
```

```python
#3.튜플(Tuple)- 인덱스로 변경 불가.
fruits = ("apple", "banana","banana" ,"strawberry","apple",)
fruits[2]="peach"
```

```python
#4.셋/집합(Set)
fruits = {"apple", "banana","banana" ,"strawberry","apple",}
print(fruits)
```

```python
#5.셋/집합(Set) - 인섹스화 불가능
fruits = {"apple", "banana","banana" ,"strawberry","apple",}
print(fruits[2])
```

```python
#6.딕셔너리/사전(Dictionary)
student = {"이름": "철수", "점수": 90,"이름": "철수", "점수": 90}
print(student["이름"])
print(student["점수"])
print(student)
student["이름"] = "나희"
student["점수"] = "100"
print(student)
```

```python
#7. 딕셔너리/사전(Dictionary) - 추가
student = {
    "이름": "철수",
    "점수": 90
}

student["학년"] = 2
student["학교"] = "한국고등학교"

print(student)
```

## 1. Python의 4가지 대표 그룹
| 자료구조 | 순서 | 수정 | 중복 |키밸류|
|---|---|---|---|-|
| 리스트 | O | O | O | X |
| 튜플 | O | X | O | X |
| 셋 | X | O | X | X |
| 딕셔너리 | O | O | X | O|

## 2. 예시 코드 이해

- [x] _리스트(List)_ 는 순서가 있고, 인덱스로 요소를 수정할 수 있으며 중복을 허용합니다.

- [x] *튜플(Tuple)* 은 순서가 있고, 인덱스로 요소를 확인할 수 있지만 수정할 수 없습니다.

- [x] ~~셋(Set)~~ 은 중복을 허용하지 않고, 인덱스로 요소에 접근할 수 없습니다.

- [x] **딕셔너리(Dictionary)** 는 키와 밸류가 한 쌍으로 이루어져 있으며, 키를 이용해 밸류를 찾을 수 있습니다.
