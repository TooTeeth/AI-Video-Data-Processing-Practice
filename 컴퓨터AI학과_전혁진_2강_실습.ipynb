## 3. 이미지 프로세싱 기초
### 3-1. 관심 영역(ROI) 추출 및 색상 변경



```python
x, y = w // 4, h // 4
roi_width, roi_height = max(1, w // 2), max(1, h // 2)
x2, y2 = min(w, x + roi_width), min(h, y + roi_height)
roi_width, roi_height = x2 - x, y2 - y
roi = image[y:y2, x:x2].copy()
print(f'ROI 시작 (x, y)=({x}, {y}), 너비={roi_width}, 높이={roi_height}')
print(f'슬라이싱: image[{y}:{y2}, {x}:{x2}]')
show_cv('중앙 ROI만 추출', roi)

roi_filled = image.copy()
roi_filled[y:y2, x:x2] = (255, 0, 0)  # OpenCV BGR: 파란색
show_cv('중앙 ROI를 파란색으로 변경', roi_filled)
results['03_roi_crop'] = roi
results['04_roi_filled'] = roi_filled
```

### 3-2. 컬러·그레이스케일 읽기 및 변환



```python
image_color = cv2.imread(str(WORK_IMAGE_PATH), cv2.IMREAD_COLOR)
image_gray_read = cv2.imread(str(WORK_IMAGE_PATH), cv2.IMREAD_GRAYSCALE)
if image_color is None or image_gray_read is None:
    raise OSError('작업용 PNG를 다시 읽지 못했습니다.')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
print('컬러 shape:', image_color.shape)
print('흑백 shape:', image_gray_read.shape)
show_cv('흑백으로 읽은 사진', image_gray_read)
show_cv('cvtColor로 변환한 흑백 사진', gray)
results['05_grayscale'] = gray
```

### 3-3. 채널 분리와 합치기



```python
B, G, R = cv2.split(image)
for channel_name, channel in [('R', R), ('G', G), ('B', B)]:
    show_cv(f'{channel_name} 채널의 세기', channel)
merged_bgr = cv2.merge([B, G, R])
merge_equal = np.array_equal(image, merged_bgr)
print('분리 후 다시 합친 이미지가 원본과 같은가?', merge_equal)
show_cv('BGR 순서로 다시 합친 사진', merged_bgr)
results['06_channel_R'] = R
results['07_channel_G'] = G
results['08_channel_B'] = B
results['09_merged'] = merged_bgr
```

### 3-4. 분리한 채널을 각 색상으로 표현하기



```python
zeros = np.zeros((h, w), dtype=np.uint8)
red_only = cv2.merge([zeros, zeros, R])
green_only = cv2.merge([zeros, G, zeros])
blue_only = cv2.merge([B, zeros, zeros])
show_cv('R 성분만 남긴 사진', red_only)
show_cv('G 성분만 남긴 사진', green_only)
show_cv('B 성분만 남긴 사진', blue_only)
results['10_red_only'] = red_only
results['11_green_only'] = green_only
results['12_blue_only'] = blue_only
```

### 3-5. 다른 색공간으로 변환하기

 **각 색공간의 세 채널을 왼쪽부터 순서대로 흑백으로 표시**한다.


```python
rgba = cv2.cvtColor(image, cv2.COLOR_BGR2RGBA)
color_spaces = {
    'HSV': cv2.cvtColor(image, cv2.COLOR_BGR2HSV),
    'Lab': cv2.cvtColor(image, cv2.COLOR_BGR2LAB),
    'Luv': cv2.cvtColor(image, cv2.COLOR_BGR2LUV),
    'YUV': cv2.cvtColor(image, cv2.COLOR_BGR2YUV),
}
space_info = []
for name, arr in [('BGR', image), ('RGB', rgb), ('RGBA', rgba), ('GRAY', gray), *color_spaces.items()]:
    space_info.append({'색공간': name, 'shape': str(arr.shape), 'dtype': str(arr.dtype)})
display(pd.DataFrame(space_info))

for name, arr in color_spaces.items():
    # 원시 3채널 배열을 컬러 사진으로 오해하지 않도록 단일 채널 3개로 출력한다.
    components = cv2.hconcat(list(cv2.split(arr)))
    show_cv(f'{name}: 첫째 / 둘째 / 셋째 채널 (흑백)', components)

print('RGBA 알파 채널 값:', np.unique(rgba[:, :, 3]))
plt.figure(figsize=(6, 4))
plt.imshow(rgba)
plt.title('RGBA image (alpha = 255)')
plt.axis('off')
plt.tight_layout()
plt.show()
plt.close()
```

### 4-1. 선, 사각형, 원, 타원, 다각형



```python
canvas = np.full((400, 600, 3), 255, dtype=np.uint8)
red, green, blue = (0, 0, 255), (0, 180, 0), (255, 0, 0)
cv2.line(canvas, (25, 40), (570, 40), red, 4, cv2.LINE_AA)
cv2.rectangle(canvas, (30, 80), (180, 190), blue, 3)
cv2.rectangle(canvas, (210, 80), (330, 190), green, -1)
cv2.circle(canvas, (450, 140), 60, red, 3, cv2.LINE_AA)
cv2.ellipse(canvas, (160, 290), (100, 45), 20, 0, 360, green, 3, cv2.LINE_AA)
triangle = np.array([[370, 230], [320, 355], [530, 355]], dtype=np.int32)
cv2.polylines(canvas, [triangle], True, blue, 3, cv2.LINE_AA)
show_cv('선과 도형 그리기', canvas)
results['13_drawing_shapes'] = canvas
```

### 4-2. 글씨 쓰기


```python
text_canvas = np.full((230, 620, 3), 255, dtype=np.uint8)
cv2.putText(text_canvas, 'Image Processing Practice', (20, 65),
            cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 0), 2, cv2.LINE_AA)
cv2.putText(text_canvas, 'OpenCV + NumPy', (20, 125),
            cv2.FONT_HERSHEY_DUPLEX, 1, blue, 2, cv2.LINE_AA)
cv2.putText(text_canvas, 'My Photo, My Practice', (20, 185),
            cv2.FONT_HERSHEY_SCRIPT_SIMPLEX, 1, red, 2, cv2.LINE_AA)
show_cv('글씨 쓰기', text_canvas)
results['14_drawing_text'] = text_canvas
```

### 4-3. 내 사진 위에 ROI와 중심 위치 표시하기



```python
annotated = image.copy()
thickness = max(1, min(h, w) // 150)
cv2.rectangle(annotated, (x, y), (x2 - 1, y2 - 1), (0, 255, 255), thickness)
cv2.circle(annotated, (center_x, center_y), max(1, min(h, w) // 25),
            (0, 0, 255), thickness, cv2.LINE_AA)
font_scale = max(0.25, min(1.5, w / 700))
label = 'CENTER ROI'
(label_w, _), _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, font_scale, thickness)
font_scale *= min(1.0, max(1, w - 2) / max(1, label_w))
text_x = max(0, min(x, w - max(1, round(label_w * min(1.0, max(1, w - 2) / max(1, label_w))))))
text_y = max(0, min(h - 1, y - 8 if y > 25 else y + 20))
cv2.putText(annotated, label, (text_x, text_y), cv2.FONT_HERSHEY_SIMPLEX,
            font_scale, (0, 255, 255), thickness, cv2.LINE_AA)
show_cv('내 사진에 ROI와 중심점 표시', annotated)
results['15_photo_annotated'] = annotated
```
